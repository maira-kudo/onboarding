```python
import rasterio
import numpy as np

def mask_raster_by_threshold(raster_path, threshold, output_path):
    """
    Cria um raster binário (0/1) baseado em um threshold e preserva metadados.
    
    Args:
        raster_path (str): Caminho para o raster de entrada (.tif).
        threshold (float): Valor de corte. Pixels > threshold viram 1, 
                           senão 0.
        output_path (str): Caminho para salvar o raster binário.
    """

    with rasterio.open(raster_path) as src:
        banda1 = src.read(1)
        meta = src.meta.copy()
        
        binary_mask = np.where(banda1 > threshold, 1, 0).astype(np.uint8)
        
        # Atualiza os metadados para o novo raster
        meta.update({
            'dtype': 'uint8',
            'count': 1,
            'nodata': 0  
        })
        
        # Grava o arquivo de saída
        with rasterio.open(output_path, 'w', **meta) as dst:
            dst.write(binary_mask, 1)
            
    print(f"Raster binário salvo em: {output_path}")

mask_raster_by_threshold(r'/home/maira/anaconda3/envs/env_onboarding/notebooks/Orthos/Orthomosaico.tif', 1500, 'raster_binario.tif')
```
