```python
import os
import zipfile
import geopandas as gpd

def des_zip(): 
    pasta_destino = r'Vectors'
    arquivo_zip = r'/home/maira/anaconda3/envs/env_onboarding/notebooks/onboarding/vectors_case_1.zip'
    with zipfile.ZipFile(arquivo_zip, 'r') as zip_ref:
        zip_ref.extractall(pasta_destino)
```

```python
root = r'/home/maira/anaconda3/envs/env_onboarding/notebooks/onboarding/Vectors/vectors_case_1'


def load_vector(path):
    files = os.listdir(root)
    metas = [] 
    for file in files: 
        gdf = gpd.read_file(os.path.join(root,file))
        meta = {
            'name': file, 
            'crs': gdf.crs,
            'poligno': gdf.geom_type.unique(), 
            'dtypes': gdf.dtypes.to_dict()
            
        }
        metas.append(meta) 
    return metas

metas = load_vector(root)
```
