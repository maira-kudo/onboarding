```python
root = r'/home/maira/anaconda3/envs/env_onboarding/notebooks/onboarding/Vectors/vectors_case_1'

def load_vector(path):
    files = os.listdir(root)
    metas = [] 
    for file in files: 
        gdf = gpd.read_file(os.path.join(root,file))
        meta = {
            'name': file, 
            'crs': gdf.crs,
            'poligno': gdf.geom_type.unique(), 
            'dtypes': gdf.dtypes.to_dict()
            
        }
        metas.append(meta) 
    return metas

metas = load_vector(root)
```

```python
import geopandas as gpd
import math 

def identify_utm_crs(gdf): 
    if gdf.crs is None:
        raise ValueError('geodf sem CRS')
    gdf = gdf.to_crs('EPSG: 4326')
    centroid = gdf.unary_union.centroid
    longitude = centroid.x
    latitude = centroid.y

    zone = math.floor((longitude + 180) / 6) + 1

    if latitude >=0:
        epsg = 32600 + zone
        hemisfere = "Norte"

    else: 
        epsg = 32700 + zone
        hemisfere = 'Sul'
    return {
        'longitude': longitude,
        'latitude': latitude,
        'utm_zome': zone,
        'hemisferio': hemisfere,
        'epsg_manual': epsg
    }


gdf1 = gpd.read_file(r'/home/maira/anaconda3/envs/env_onboarding/notebooks/onboarding/Vectors/vectors_case_1/MATOLOGIA_Orthomosaico.geojson')
gdf2 = gpd.read_file(r'/home/maira/anaconda3/envs/env_onboarding/notebooks/onboarding/Vectors/vectors_case_1/shape.geojson')

def compare_crs(gdf, nome): 
    manual = identify_utm_crs(gdf)
    auto = gdf.estimate_utm_crs()
    print(f'Resultado manual: {manual}')
    print(f'Resultado auto: {auto}')

compare_crs(gdf1, "MATOLOGIA_Orthomosaico")
compare_crs(gdf2, "shape")

```
