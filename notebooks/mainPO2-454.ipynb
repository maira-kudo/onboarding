## SUBTASK 1 (PO2-456): 
    Calcular comprimento de geometrias invalidas

```python
import math
import geopandas as gpd

gdf1 = gpd.read_file(r'/home/maira/anaconda3/envs/env_onboarding/notebooks/onboarding/Vectors/vectors_case_1/MATOLOGIA_Orthomosaico.geojson')
gdf2 = gpd.read_file(r'/home/maira/anaconda3/envs/env_onboarding/notebooks/onboarding/Vectors/vectors_case_1/shape.geojson')

def identify_utm_epsg(gdf):
    gdf_latlon = gdf.to_crs('EPSG: 4326')
    centroid = gdf_latlon.unary_union.centroid

    lon = centroid.x
    lat = centroid.y

    zone = math.floor((lon + 180) / 6)+1

    if lat >= 0:
        return 32600 + zone
    else: 
        return 327000 + zone 


def add_linestring_length(gdf, col_name='length_m'): 
    target_epsg = identify_utm_epsg(gdf)
    gdf_projected = gdf.to_crs(epsg = target_epsg)
    gdf_projected[col_name] = gdf_projected.geometry.length
    return gdf_projected

print(gdf1.crs)
gdf_result = add_linestring_length(gdf1)

print(gdf_result[['geometry', 'length_m']].head())
gdf_result.to_file('linhas_com_comprimento.geojson', driver="GeoJSON")
```

## SUBTASK 2 (PO2-455): 

Calcular área de poligonos 

```python
def identify_utm_epsg(gdf):
    gdf_latlon = gdf.to_crs('EPSG: 4326')
    centroid = gdf_latlon.unary_union.centroid

    lon = centroid.x
    lat = centroid.y

    zone = math.floor((lon + 180) / 6)+1

    if lat >= 0:
        return 32600 + zone
    else: 
        return 327000 + zone 
    

def add_polygon_area(gdf, col_name='area_ha'): 
    target_epsg = identify_utm_epsg(gdf)
    gdf_projected = gdf.to_crs(epsg = target_epsg)
    area_m2 = gdf_projected.geometry.area
    gdf_projected[col_name] = area_m2 / 10000
    return gdf_projected

print(gdf1.crs)


gdf_result = add_polygon_area(gdf1)

print(gdf_result[['geometry', 'area_ha']].head())
gdf_result.to_file('poligonos_area.geojson', driver="GeoJSON")
```

## SUBTASK 3 (PO2-458): 

Explodir geometrias multipart

```python
def explode_multipart(gdf):
    exploded = gdf.explode(index_parts = False).reset_index(drop=True)
    return exploded

gdf1.geometry.geom_type.value_counts()
gdf_exploded = explode_multipart(gdf1)

gdf_exploded.geometry.geom_type.value_counts()

print(f'len original: {len(gdf1)}')
print(f'len exploded: {len(gdf_exploded)}')

```
